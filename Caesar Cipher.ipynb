```python

def Caesar_cipher_encryptV2(plainteks, k):
    cipherteks = ""
    for char in plainteks:
        c = (ord(char) + k) % 256
        cipherteks = cipherteks + chr(c)
    return cipherteks

```

```python
def Caesar_cipher_decryptV2(cipherteks, k):
    plainteks = ""
    for char in cipherteks:
        c = (ord(char) - k) % 256
        plainteks = plainteks + chr(c)
    return plainteks

```

```python
pesan = input("ketikkan pesan anda: ")

kunci = int(input("kunci (1-255): "))

cipherteks = Caesar_cipher_encryptV2(pesan, kunci)

print(f"Pesan terenkripsi: {cipherteks}")
```

```python
plainteks = Caesar_cipher_decryptV2(cipherteks, kunci)

print(f"Pesan hasil dekripsi: {plainteks}")
```
